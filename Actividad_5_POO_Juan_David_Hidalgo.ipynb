Ejercicio 8.4: Cuadros de dialogo

```python
import os
import tkinter as tk
from tkinter import ttk, messagebox, filedialog

# Clases del modelo de negocio

class Empleado:
    def __init__(self, nombre: str, apellidos: str, cargo: str, genero: str,
                 salario_dia: float, dias_trabajados: int, otros_ingresos: float,
                 pagos_salud: float, aporte_pensiones: float):
        self.nombre = nombre
        self.apellidos = apellidos
        self.cargo = cargo
        self.genero = genero
        self.salario_dia = salario_dia
        self.dias_trabajados = dias_trabajados
        self.otros_ingresos = otros_ingresos
        self.pagos_salud = pagos_salud
        self.aporte_pensiones = aporte_pensiones

    def calcular_nomina(self) -> float:
        return (self.dias_trabajados * self.salario_dia) + self.otros_ingresos - self.pagos_salud - self.aporte_pensiones


class ListaEmpleados:
    def __init__(self):
        self.lista = []

    def agregar_empleado(self, empleado: Empleado):
        self.lista.append(empleado)

    def calcular_total_nomina(self) -> float:
        return sum(emp.calcular_nomina() for emp in self.lista)

    def obtener_matriz_datos(self):
        matriz = []
        for emp in self.lista:
            matriz.append((emp.nombre, emp.apellidos, f"{emp.calcular_nomina():.2f}"))
        return matriz

    def convertir_texto(self) -> str:
        texto = ""
        for emp in self.lista:
            texto += (
                f"Nombre = {emp.nombre}\n"
                f"Apellidos = {emp.apellidos}\n"
                f"Cargo = {emp.cargo}\n"
                f"Género = {emp.genero}\n"
                f"Salario = ${emp.salario_dia:.2f}\n"
                f"Días trabajados = {emp.dias_trabajados}\n"
                f"Otros ingresos = ${emp.otros_ingresos:.2f}\n"
                f"Pagos salud = ${emp.pagos_salud:.2f}\n"
                f"Aportes pensiones = ${emp.aporte_pensiones:.2f}\n"
                f"Sueldo mensual = ${emp.calcular_nomina():.2f}\n"
                f"---------\n"
            )
        texto += f"Total nómina = ${self.calcular_total_nomina():.2f}\n"
        return texto


# Interfaces gráficas de usuario y cuadros de diálogo

class VentanaAgregarEmpleado(tk.Toplevel):
    def __init__(self, parent, lista_empleados: ListaEmpleados):
        super().__init__(parent)
        self.title("Agregar Empleado")
        self.geometry("350x440")
        self.resizable(False, False)
        self.lista_empleados = lista_empleados

        self.transient(parent)
        self.grab_set()

        frame = tk.Frame(self)
        frame.pack(padx=15, pady=15, fill="both", expand=True)

        # Nombre
        tk.Label(frame, text="Nombre:").grid(row=0, column=0, sticky="w", pady=4)
        self.txt_nombre = tk.Entry(frame, width=22)
        self.txt_nombre.grid(row=0, column=1, pady=4)

        # Apellidos
        tk.Label(frame, text="Apellidos:").grid(row=1, column=0, sticky="w", pady=4)
        self.txt_apellidos = tk.Entry(frame, width=22)
        self.txt_apellidos.grid(row=1, column=1, pady=4)

        # Cargo
        tk.Label(frame, text="Cargo:").grid(row=2, column=0, sticky="nw", pady=4)
        self.lst_cargo = tk.Listbox(frame, height=3, width=22, exportselection=False)
        for c in ["Directivo", "Estratégico", "Operativo"]:
            self.lst_cargo.insert(tk.END, c)
        self.lst_cargo.select_set(0)
        self.lst_cargo.grid(row=2, column=1, pady=4)

        # Género
        tk.Label(frame, text="Género:").grid(row=3, column=0, sticky="w", pady=4)
        frame_gen = tk.Frame(frame)
        frame_gen.grid(row=3, column=1, sticky="w", pady=4)
        self.var_genero = tk.StringVar(value="Masculino")
        tk.Radiobutton(frame_gen, text="Masculino", variable=self.var_genero, value="Masculino").pack(side="left")
        tk.Radiobutton(frame_gen, text="Femenino", variable=self.var_genero, value="Femenino").pack(side="left")

        # Salario por día
        tk.Label(frame, text="Salario por día:").grid(row=4, column=0, sticky="w", pady=4)
        self.txt_salario_dia = tk.Entry(frame, width=22)
        self.txt_salario_dia.grid(row=4, column=1, pady=4)

        # Días trabajados
        tk.Label(frame, text="Días trabajados:").grid(row=5, column=0, sticky="w", pady=4)
        self.spn_dias = tk.Spinbox(frame, from_=1, to=31, width=20)
        self.spn_dias.grid(row=5, column=1, pady=4)

        # Otros ingresos
        tk.Label(frame, text="Otros ingresos:").grid(row=6, column=0, sticky="w", pady=4)
        self.txt_otros = tk.Entry(frame, width=22)
        self.txt_otros.grid(row=6, column=1, pady=4)

        # Pagos por salud
        tk.Label(frame, text="Pagos por salud:").grid(row=7, column=0, sticky="w", pady=4)
        self.txt_salud = tk.Entry(frame, width=22)
        self.txt_salud.grid(row=7, column=1, pady=4)

        # Aporte pensiones
        tk.Label(frame, text="Aporte pensiones:").grid(row=8, column=0, sticky="w", pady=4)
        self.txt_pensiones = tk.Entry(frame, width=22)
        self.txt_pensiones.grid(row=8, column=1, pady=4)

        frame_btns = tk.Frame(frame)
        frame_btns.grid(row=9, column=0, columnspan=2, pady=15)
        tk.Button(frame_btns, text="Agregar", command=self.agregar, width=12, bg="#d4edda").pack(side="left", padx=5)
        tk.Button(frame_btns, text="Limpiar", command=self.limpiar_campos, width=12, bg="#f8d7da").pack(side="left", padx=5)

    def limpiar_campos(self):
        self.txt_nombre.delete(0, tk.END)
        self.txt_apellidos.delete(0, tk.END)
        self.lst_cargo.select_clear(0, tk.END)
        self.lst_cargo.select_set(0)
        self.var_genero.set("Masculino")
        self.txt_salario_dia.delete(0, tk.END)
        self.spn_dias.delete(0, tk.END)
        self.spn_dias.insert(0, "1")
        self.txt_otros.delete(0, tk.END)
        self.txt_salud.delete(0, tk.END)
        self.txt_pensiones.delete(0, tk.END)

    def agregar(self):
        try:
            nombre = self.txt_nombre.get().strip()
            apellidos = self.txt_apellidos.get().strip()

            if not nombre or not apellidos:
                raise ValueError("El nombre y los apellidos son obligatorios.")

            seleccion_cargo = self.lst_cargo.curselection()
            if not seleccion_cargo:
                raise ValueError("Debe seleccionar un cargo.")
            cargo = self.lst_cargo.get(seleccion_cargo[0])
            genero = self.var_genero.get()

            salario_dia = float(self.txt_salario_dia.get())
            dias_trabajados = int(self.spn_dias.get())
            if dias_trabajados < 1 or dias_trabajados > 31:
                raise ValueError("Los días trabajados deben estar entre 1 y 31.")

            otros_ingresos = float(self.txt_otros.get())
            pagos_salud = float(self.txt_salud.get())
            aporte_pensiones = float(self.txt_pensiones.get())

            emp = Empleado(nombre, apellidos, cargo, genero, salario_dia,
                           dias_trabajados, otros_ingresos, pagos_salud, aporte_pensiones)
            self.lista_empleados.agregar_empleado(emp)

            messagebox.showinfo("Mensaje", "El empleado ha sido agregado", parent=self)
            self.limpiar_campos()

        except ValueError as e:
            messagebox.showerror("Error", f"Campo nulo o error en formato de número:\n{e}", parent=self)


class VentanaNomina(tk.Toplevel):
    def __init__(self, parent, lista_empleados: ListaEmpleados):
        super().__init__(parent)
        self.title("Nómina de Empleados")
        self.geometry("420x320")
        self.resizable(False, False)
        self.transient(parent)
        self.grab_set()

        tk.Label(self, text="Detalle de Nómina Mensual", font=("Arial", 11, "bold")).pack(pady=10)

        columnas = ("Nombres", "Apellidos", "Sueldo")
        self.tabla = ttk.Treeview(self, columns=columnas, show="headings", height=8)

        for col in columnas:
            self.tabla.heading(col, text=col)
            self.tabla.column(col, width=125, anchor="center")

        for fila in lista_empleados.obtener_matriz_datos():
            self.tabla.insert("", tk.END, values=fila)

        self.tabla.pack(padx=15, pady=5)

        total = lista_empleados.calcular_total_nomina()
        tk.Label(self, text=f"Total nómina mensual = $ {total:.2f}",
                 font=("Arial", 10, "bold"), fg="darkgreen").pack(pady=15)


class VentanaPrincipal(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("Nómina")
        self.geometry("320x360")
        self.resizable(False, False)

        self.empleados = ListaEmpleados()

        # Barra de Menús
        barra_menu = tk.Menu(self)
        menu_opciones = tk.Menu(barra_menu, tearoff=0)

        menu_opciones.add_command(label="Agregar empleado", command=self.abrir_agregar_empleado)
        menu_opciones.add_command(label="Calcular nómina", command=self.abrir_calcular_nomina)
        menu_opciones.add_separator()
        menu_opciones.add_command(label="Guardar archivo", command=self.guardar_archivo)

        barra_menu.add_cascade(label="Menú", menu=menu_opciones)
        self.config(menu=barra_menu)

        tk.Label(self, text="Sistema de Gestión de Nómina", font=("Arial", 12, "bold")).pack(pady=60)
        tk.Label(self, text="Utilice la barra de menú superior\npara gestionar los empleados,\ncalcular la nómina y exportar\nel archivo 'Nómina.txt'.",
                 font=("Arial", 10), justify="center", fg="#444444").pack(pady=10)

    def abrir_agregar_empleado(self):
        VentanaAgregarEmpleado(self, self.empleados)

    def abrir_calcular_nomina(self):
        VentanaNomina(self, self.empleados)

    def guardar_archivo(self):
        carpeta = filedialog.askdirectory(title="Seleccione la carpeta para guardar Nómina.txt")
        if carpeta:
            ruta_archivo = os.path.join(carpeta, "Nómina.txt")
            try:
                with open(ruta_archivo, "w", encoding="utf-8") as archivo:
                    archivo.write(self.empleados.convertir_texto())
                messagebox.showinfo("Mensaje", f"El archivo de nómina Nómina.txt se ha creado en:\n{carpeta}")
            except Exception as e:
                messagebox.showerror("Error", f"No se pudo guardar el archivo:\n{e}")


if __name__ == "__main__":
    try:
        app = VentanaPrincipal()
        app.mainloop()
    except Exception as e:
        print(f"Entorno sin interfaz gráfica o error de ejecución: {e}")
```

Ejercicio 8.5: Gestion de Contenido

```python
from datetime import datetime
import tkinter as tk
from tkinter import ttk, messagebox

#Clases del modelo de negocio

class Huesped:
    def __init__(self, nombres: str, apellidos: str, documento: str, fecha_ingreso: datetime):
        self.nombres = nombres
        self.apellidos = apellidos
        self.documento = documento
        self.fecha_ingreso = fecha_ingreso


class Habitacion:
    def __init__(self, numero: int, precio_dia: float):
        self.numero = numero
        self.precio_dia = precio_dia
        self.disponible = True
        self.huesped = None

    def ocupar(self, huesped: Huesped):
        self.huesped = huesped
        self.disponible = False

    def liberar(self):
        self.huesped = None
        self.disponible = True

    def calcular_dias_alojamiento(self, fecha_salida: datetime) -> int:
        if not self.huesped:
            return 0
        dias = (fecha_salida - self.huesped.fecha_ingreso).days
        return dias

    def calcular_total_pagar(self, fecha_salida: datetime) -> float:
        dias = self.calcular_dias_alojamiento(fecha_salida)
        return dias * self.precio_dia


class Hotel:
    def __init__(self):
        self.habitaciones = []
        for i in range(1, 11):
            precio = 120000.0 if i <= 5 else 160000.0
            self.habitaciones.append(Habitacion(i, precio))

    def buscar_habitacion(self, numero: int):
        for hab in self.habitaciones:
            if hab.numero == numero:
                return hab
        return None


#Interfaces graficas de usuario

class VentanaRegistroHuesped(tk.Toplevel):
    def __init__(self, parent, habitacion: Habitacion, ventana_consulta):
        super().__init__(parent)
        self.title(f"Ingreso - Habitación {habitacion.numero}")
        self.geometry("360x280")
        self.resizable(False, False)
        self.habitacion = habitacion
        self.ventana_consulta = ventana_consulta
        self.transient(parent)
        self.grab_set()

        frame = tk.Frame(self, padx=15, pady=15)
        frame.pack(fill="both", expand=True)

        tk.Label(frame, text=f"Habitación #{habitacion.numero} (${habitacion.precio_dia:,.0f}/día)",
                 font=("Arial", 10, "bold")).grid(row=0, column=0, columnspan=2, pady=(0, 10))

        tk.Label(frame, text="Fecha ingreso (AAAA-MM-DD):").grid(row=1, column=0, sticky="w", pady=4)
        self.txt_fecha = tk.Entry(frame, width=18)
        self.txt_fecha.insert(0, datetime.now().strftime("%Y-%m-%d"))
        self.txt_fecha.grid(row=1, column=1, pady=4)

        tk.Label(frame, text="Nombres:").grid(row=2, column=0, sticky="w", pady=4)
        self.txt_nombres = tk.Entry(frame, width=18)
        self.txt_nombres.grid(row=2, column=1, pady=4)

        tk.Label(frame, text="Apellidos:").grid(row=3, column=0, sticky="w", pady=4)
        self.txt_apellidos = tk.Entry(frame, width=18)
        self.txt_apellidos.grid(row=3, column=1, pady=4)

        tk.Label(frame, text="N° Documento:").grid(row=4, column=0, sticky="w", pady=4)
        self.txt_doc = tk.Entry(frame, width=18)
        self.txt_doc.grid(row=4, column=1, pady=4)

        tk.Button(frame, text="Registrar Huésped", command=self.registrar,
                  bg="#d4edda", width=20).grid(row=5, column=0, columnspan=2, pady=15)

    def registrar(self):
        fecha_str = self.txt_fecha.get().strip()
        nombres = self.txt_nombres.get().strip()
        apellidos = self.txt_apellidos.get().strip()
        doc = self.txt_doc.get().strip()

        if not fecha_str or not nombres or not apellidos or not doc:
            messagebox.showerror("Error", "Todos los campos son obligatorios.", parent=self)
            return

        try:
            fecha_ingreso = datetime.strptime(fecha_str, "%Y-%m-%d")
        except ValueError:
            messagebox.showerror("Error de formato", "Use el formato de fecha AAAA-MM-DD (ej. 2026-11-20).", parent=self)
            return

        huesped = Huesped(nombres, apellidos, doc, fecha_ingreso)
        self.habitacion.ocupar(huesped)
        messagebox.showinfo("Registro exitoso", f"Habitación {self.habitacion.numero} ocupada por {nombres} {apellidos}.", parent=self)
        self.ventana_consulta.actualizar_tabla()
        self.destroy()


class VentanaConsultarHabitaciones(tk.Toplevel):
    def __init__(self, parent, hotel: Hotel):
        super().__init__(parent)
        self.title("Consultar Habitaciones")
        self.geometry("460x380")
        self.resizable(False, False)
        self.hotel = hotel
        self.transient(parent)
        self.grab_set()

        tk.Label(self, text="Estado de Habitaciones (1 - 10)", font=("Arial", 11, "bold")).pack(pady=8)

        columnas = ("Habitación", "Precio/Día", "Estado")
        self.tabla = ttk.Treeview(self, columns=columnas, show="headings", height=10)
        for col in columnas:
            self.tabla.heading(col, text=col)
            self.tabla.column(col, width=130, anchor="center")
        self.tabla.pack(padx=15, pady=5)

        self.actualizar_tabla()

        frame_sel = tk.Frame(self, pady=10)
        frame_sel.pack()

        tk.Label(frame_sel, text="Número de habitación a ocupar:").pack(side="left", padx=5)
        self.txt_num_hab = tk.Entry(frame_sel, width=6)
        self.txt_num_hab.pack(side="left", padx=5)
        tk.Button(frame_sel, text="Ocupar", command=self.seleccionar_habitacion, bg="#d1ecf1").pack(side="left", padx=5)

    def actualizar_tabla(self):
        for item in self.tabla.get_children():
            self.tabla.delete(item)
        for hab in self.hotel.habitaciones:
            estado = "Disponible" if hab.disponible else "No disponible"
            self.tabla.insert("", tk.END, values=(f"Habitación {hab.numero}", f"${hab.precio_dia:,.0f}", estado))

    def seleccionar_habitacion(self):
        num_str = self.txt_num_hab.get().strip()
        if not num_str.isdigit():
            messagebox.showerror("Error", "Ingrese un número de habitación válido.", parent=self)
            return

        hab = self.hotel.buscar_habitacion(int(num_str))
        if not hab:
            messagebox.showerror("Error", "La habitación no existe (debe ser entre 1 y 10).", parent=self)
            return

        if not hab.disponible:
            messagebox.showerror("Habitación ocupada", f"La habitación {hab.numero} no está disponible.", parent=self)
            return

        VentanaRegistroHuesped(self, hab, self)


class VentanaProcesoSalida(tk.Toplevel):
    def __init__(self, parent, habitacion: Habitacion):
        super().__init__(parent)
        self.title(f"Salida - Habitación {habitacion.numero}")
        self.geometry("380x320")
        self.resizable(False, False)
        self.habitacion = habitacion
        self.transient(parent)
        self.grab_set()

        huesped = habitacion.huesped
        fecha_ing_str = huesped.fecha_ingreso.strftime("%Y-%m-%d")

        frame = tk.Frame(self, padx=15, pady=15)
        frame.pack(fill="both", expand=True)

        tk.Label(frame, text=f"Entrega de Habitación #{habitacion.numero}", font=("Arial", 10, "bold")).pack(pady=(0, 5))
        tk.Label(frame, text=f"Huésped: {huesped.nombres} {huesped.apellidos}").pack(anchor="w")
        tk.Label(frame, text=f"Fecha de ingreso: {fecha_ing_str}").pack(anchor="w", pady=(0, 10))

        frame_fecha = tk.Frame(frame)
        frame_fecha.pack(fill="x", pady=5)
        tk.Label(frame_fecha, text="Fecha salida (AAAA-MM-DD):").pack(side="left")
        self.txt_fecha_salida = tk.Entry(frame_fecha, width=14)
        self.txt_fecha_salida.pack(side="left", padx=5)

        self.btn_calcular = tk.Button(frame, text="Calcular Días y Total a Pagar", command=self.calcular_totales, bg="#fff3cd")
        self.btn_calcular.pack(pady=8)

        self.lbl_resumen = tk.Label(frame, text="Días de alojamiento: - | Total a pagar: -", font=("Arial", 9, "bold"), fg="blue")
        self.lbl_resumen.pack(pady=5)

        self.btn_registrar_salida = tk.Button(frame, text="Registrar salida", command=self.registrar_salida,
                                              bg="#d4edda", state=tk.DISABLED, width=20)
        self.btn_registrar_salida.pack(pady=10)

    def calcular_totales(self):
        salida_str = self.txt_fecha_salida.get().strip()
        try:
            fecha_salida = datetime.strptime(salida_str, "%Y-%m-%d")
        except ValueError:
            messagebox.showerror("Error de fecha", "Ingrese una fecha válida en formato AAAA-MM-DD.", parent=self)
            return

        if fecha_salida <= self.habitacion.huesped.fecha_ingreso:
            messagebox.showerror("Fecha inválida", "La fecha de salida debe ser mayor a la fecha de ingreso.", parent=self)
            self.btn_registrar_salida.config(state=tk.DISABLED)
            return

        dias = self.habitacion.calcular_dias_alojamiento(fecha_salida)
        total = self.habitacion.calcular_total_pagar(fecha_salida)
        self.lbl_resumen.config(text=f"Días de alojamiento: {dias} | Total a pagar: ${total:,.0f}", fg="darkgreen")
        self.btn_registrar_salida.config(state=tk.NORMAL)

    def registrar_salida(self):
        self.habitacion.liberar()
        messagebox.showinfo("Salida registrada", f"La habitación {self.habitacion.numero} ahora está disponible.", parent=self)
        self.destroy()


class VentanaSalidaHuesped(tk.Toplevel):
    def __init__(self, parent, hotel: Hotel):
        super().__init__(parent)
        self.title("Salida de Huéspedes")
        self.geometry("320x170")
        self.resizable(False, False)
        self.hotel = hotel
        self.transient(parent)
        self.grab_set()

        tk.Label(self, text="Salida de Huéspedes", font=("Arial", 11, "bold")).pack(pady=15)

        frame = tk.Frame(self)
        frame.pack(pady=5)
        tk.Label(frame, text="Número de habitación:").pack(side="left", padx=5)
        self.txt_numero = tk.Entry(frame, width=8)
        self.txt_numero.pack(side="left", padx=5)

        tk.Button(self, text="Continuar", command=self.validar_habitacion, bg="#d1ecf1", width=15).pack(pady=15)

    def validar_habitacion(self):
        num_str = self.txt_numero.get().strip()
        if not num_str.isdigit():
            messagebox.showerror("Error", "Ingrese un número de habitación válido.", parent=self)
            return

        hab = self.hotel.buscar_habitacion(int(num_str))
        if not hab:
            messagebox.showerror("Error", "La habitación ingresada no existe.", parent=self)
            return

        if hab.disponible:
            messagebox.showerror("Habitación disponible", f"La habitación {hab.numero} no está ocupada actualmente.", parent=self)
            return

        VentanaProcesoSalida(self, hab)
        self.destroy()


class VentanaPrincipalHotel(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("Gestión de Hotel")
        self.geometry("340x320")
        self.resizable(False, False)

        self.hotel = Hotel()

        barra_menu = tk.Menu(self)
        menu_hotel = tk.Menu(barra_menu, tearoff=0)
        menu_hotel.add_command(label="Consultar habitaciones", command=self.abrir_consultar)
        menu_hotel.add_command(label="Salida de huéspedes", command=self.abrir_salida)
        barra_menu.add_cascade(label="Menú", menu=menu_hotel)
        self.config(menu=barra_menu)

        tk.Label(self, text="Sistema de Gestión Hotelera", font=("Arial", 12, "bold")).pack(pady=50)
        tk.Label(self, text="Habitaciones 1-5: $120.000 / día\nHabitaciones 6-10: $160.000 / día\n\nUse 'Menú' en la barra superior.",
                 font=("Arial", 9), justify="center", fg="#444444").pack(pady=10)

    def abrir_consultar(self):
        VentanaConsultarHabitaciones(self, self.hotel)

    def abrir_salida(self):
        VentanaSalidaHuesped(self, self.hotel)


if __name__ == "__main__":
    try:
        app = VentanaPrincipalHotel()
        app.mainloop()
    except Exception as e:
        print(f"Entorno sin interfaz gráfica o error de ejecución: {e}")
```
